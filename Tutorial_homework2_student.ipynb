### Transcription factor circuits:

In the paper "Synthetic multistability in mammalian cells" (MultiFate), the authors demonstrated that a synthetic system with multiple stable transcription factor states can be generated with a small number of self activating, dimerizing transcription factors.

Suppose that we would like to replicate the results from the MultiFate paper and are curious if we can construct our own synthetic circuits with slightly different zinc finger transcription factors.

To make sure that we are correctly modeling the system of transcription factors, let's first start by recapitulating the modeling results from the paper. The system of differential equations for transcription factor $A_{tot}$ and $B_{tot}$ are defined in the paper (from Box 1):

$d[A_{tot}] = \alpha + \frac{\beta [A_2]^n}{1+[A_2]^n} - [A_{tot}]$

$d[B_{tot}] = \alpha + \frac{\beta [B_2]^n}{1+[B_2]^n} - [B_{tot}]$

Where $\alpha$ is the constant base production rate of both transcription factors, $\beta$ is the maximal self activation rate of the transcription factor and $n$ is the Hill coefficient ([steepness of self response curve](https://en.wikipedia.org/wiki/Hill_equation_(biochemistry)#Hill_coefficient)). 

The concentrations of the homodimers are solved for by introducing a homodimer dissociation constant $K_d$ (controlling the proportion of TF monomer to dimers) and conservation mass to get:

$[A_2] = \frac{2[A_{tot}^2]}{K_d + 4 ([A_{tot}]+[B_{tot}])+\sqrt{K_d^2+8([A_{tot}]+[B_{tot}])K_d}}$

$[B_2] = \frac{2[B_{tot}^2]}{K_d + 4 ([A_{tot}]+[B_{tot}])+\sqrt{K_d^2+8([A_{tot}]+[B_{tot}])K_d}}$
